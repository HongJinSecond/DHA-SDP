```python
##### run for F1 ##########
plt.figure(figsize=(15,6))
model_name=[]
model_performance=[]
########
model_name.append("codet5")
model_performance.append(codet5_performance["f1"].values[-1].item())

model_name.append("codebert")
model_performance.append(codebert_performance["f1"].values[-1].item())

model_name.append("unixcoder")
model_performance.append(unixcoder_performance["f1"].values[-1].item())

model_name.append("graphcodebert")
model_performance.append(graphcodebert_performance["f1"].values[-1].item())

model_name.append("plbart")
model_performance.append(plbart_performance["f1"].values[-1].item())

model_name.append("CC2Vec")
model_performance.append(baseline_performance[baseline_performance["method"]=="CC2Vec"]["f1"].values[-1].item())

model_name.append("JITLine")
model_performance.append(baseline_performance[baseline_performance["method"]=="JITLine"]["f1"].values[-1].item())

model_name.append("CCT5")
model_performance.append(baseline_performance[baseline_performance["method"]=="CCT5"]["f1"].values[-1].item())

sorted_idx = np.argsort(model_performance)
sorted_names = [model_name[i] for i in sorted_idx]
sorted_values = [model_performance[i] for i in sorted_idx]

bars=plt.bar(sorted_names,sorted_values,color="#d49a64")
# 优化视觉
plt.xlabel('Models', fontsize=12)
plt.title('Experimental Results Comparison for F1', pad=20, fontsize=14)
plt.grid(axis='x', linestyle='--', alpha=0.7)

# 添加数值标签
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/4,height+0.01,  f'{height:.4f}', 
             va='center', fontsize=10)
```
